#### Snyder ODE System

$Hollings = \frac{hHM}{hηM+1}$

$Leemput = 1 - \frac{H}{(1-σ) + (σC)}$

$\frac{dC}{dt} = r_CCS + α_{CT}r_CCT - α_{MC}r_MMC - μ_C C$ 

$\frac{dM}{dt} = r_MMS + α_{MT}r_MMT + α_{MC}r_MMC - Hollings$

$\frac{dT}{dt} = r_TTS - α_{CT}r_CCT - α_{MT}r_MMT - Hollings - Hollings$

$\frac{dH}{dt} = (sH + eHollings)*Leemput + θH - γH - μH$

```python
def SnyderODE(t, N,                                         #timesteps
              rC, rM, rT, αCT, αMT, αMC, μC,                #benthos
              sB, hB, ηB, eB, σB, θB, γB, μB,               #browsers
              sG, hG, ηG, eG, σG, θG, γG, μG,               #grazers
              sS, hS, ηS, eS, σS, θS, γS, μS):              #scrapers   

    
    C, M, T, B, G, S = N

    
    Hollings_B = (hB*B*M)/((hB*ηB*M)+1)
    Hollings_G = (hG*G*T)/((hG*ηG*T)+1)
    Hollings_S = (hS*S*T)/((hS*ηS*T)+1)

    
    Leemput_B = 1-(B/((1-σB) + (σB*C)))
    Leemput_G = 1-(G/((1-σG) + (σG*C)))
    Leemput_S = 1-(S/((1-σS) + (σS*C)))

    
    dC = rC*C*(1-C-M-T) + αCT*rC*C*T - αMC*rM*M*C - μC*C
    dM = rM*M*(1-C-M-T) + αMT*rM*M*T + αMC*rM*M*C - Hollings_B    
    dT = rT*T*(1-C-M-T) - αCT*rC*C*T - αMT*rM*M*T - Hollings_G - Hollings_S

    
    dB = ((sB*B + eB*Hollings_B)*Leemput_B) + θB*B - γB*B - μB*B
    dG = ((sG*G + eG*Hollings_G)*Leemput_G) + θG*G - γG*G - μG*G
    dS = ((sS*S + eS*Hollings_S)*Leemput_S) + θS*S - γS*S  - μS*S

      
    return dC, dM, dT, dB, dG, dS
```
