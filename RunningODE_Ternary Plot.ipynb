### Running ODE

Remember to...
1. Initialize timesteps
2. Define Initial Benthic Cover
3. Define Scenario (unless it's already set to run them all)
4. Set parameter of interest
5. Change +/- sign in last function based on scaling based on if your're scaling up/down

```python
#Timesteps
dt = 1
NUMSTEPS = 1
NUMYEARS = int(NUMSTEPS/dt)
time_points = np.linspace(0, NUMYEARS, NUMSTEPS+1)


#Initial Herbivore Conditions
init_cond = {
    'C_init': 0.99
}

#Specify scenario names and the parameter of interest (be sure to change +/- signs in last function if you are scaling up/down)
runs = runs_hB
x = 'hB'
```

### rM v. μC for Each Herbivore Combination

```python
def run_SnyderODE(scenario, dictionary): #herbivore_dictionary

    now = datetime.now() 
    date = now.date() 

    
    x = ODEfunction(
            NUMYEARS               = NUMYEARS,
            time_points            = time_points,
            init_cond              = init_cond,
#            herbivore_dictionary   = herbivore_dictionary,
#            benthos_dictionary     = benthos_dictionary,
            dictionary             = dictionary,
            function               = SnyderODE
    )

    
    x.to_pickle(f"{scenario}_{date}")

#runs = ['NUTR_benthos', 'SED_benthos', 'SED_NUTR_benthos', 'pristine_benthos']
```

```python
og_val = base[x]

for scenario in runs:
    print(scenario)

    if scenario == runs[0]:
        run_SnyderODE(runs[0], base)

    if scenario == runs[1]:
        base[x] = og_val + (og_val*.10)
        run_SnyderODE(runs[1], base)

    if scenario == runs[2]:
        base[x] = og_val + (og_val*.50)
        run_SnyderODE(runs[2], base)

    if scenario == runs[3]:
        base[x] = og_val + (og_val*.90)
        run_SnyderODE(runs[3], base)
        
base[x] = og_val
```
