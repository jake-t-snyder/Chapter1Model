```python
def filter_Cp(df, year):

    df = df.copy() 
    outputs = []
    
    for i in range(len(df)):
        if len(df['C_array'][i]) > year:
            value = df['C_array'][i][year]
            outputs.append(value)
            df.loc[i, 'Cp'+str(year)] = value
        else:
            df.loc[i, 'Cp'+str(year)] = 0

    return(df)

rc_base_1 =  filter_Cp(rc_base, 1)
rc_base_5 =  filter_Cp(rc_base, 5)
rc_base_10 =  filter_Cp(rc_base, 10)

rc_base_50 =  filter_Cp(rc_base, 50)
rc_base_60 =  filter_Cp(rc_base, 60)
rc_base_70 =  filter_Cp(rc_base, 70)
rc_base_80 =  filter_Cp(rc_base, 80)
rc_base_90 =  filter_Cp(rc_base, 90)
rc_base_100 =  filter_Cp(rc_base, 100)

#rc_dwn10_1 =  filter_Cp(rc_dwn10, 1)
#rc_dwn10_5 =  filter_Cp(rc_dwn10, 5)
#rc_dwn10_10 =  filter_Cp(rc_dwn10, 10)
#rc_dwn10_50 =  filter_Cp(rc_dwn10, 50)
#rc_dwn10_100 =  filter_Cp(rc_dwn10, 100)

rc_dwn50_1 =  filter_Cp(rc_dwn50, 1)
rc_dwn50_5 =  filter_Cp(rc_dwn50, 5)
rc_dwn50_10 =  filter_Cp(rc_dwn50, 10)

rc_dwn50_50 =  filter_Cp(rc_dwn50, 50)
rc_dwn50_60 =  filter_Cp(rc_dwn50, 60)
rc_dwn50_70 =  filter_Cp(rc_dwn50, 70)
rc_dwn50_80 =  filter_Cp(rc_dwn50, 80)
rc_dwn50_90 =  filter_Cp(rc_dwn50, 90)
rc_dwn50_100 =  filter_Cp(rc_dwn50, 100)
```
